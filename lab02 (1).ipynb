### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
# ГРУПА Z_10
G = Integers(10)
print("Група Z_10")
print("Елементи групи:", list(G))
print("Порядок групи:", G.cardinality())
# 1. Вибираємо елементи a і b
a = G(3)
b = G(7)
print("\na =", a)
print("b =", b)
# 2. Обчислюємо a * b
# У Z_10 операція групи — додавання
print("\na + b =", a + b)
# 3. Обернені елементи
a_inv = -a
b_inv = -b
print("\na^(-1) =", a_inv)
print("b^(-1) =", b_inv)
# Перевірка
print("\nПеревірка:")
print("a + a^(-1) =", a + a_inv)
print("b + b^(-1) =", b + b_inv)
# 4. Підгрупа, породжена a і b
# Утворюємо всі суми k*a + l*b у Z_10.
H = sorted({G(k*a + l*b) for k in range(10) for l in range(10)}, key=int)
print("\nПідгрупа <a,b> =", H)
print("Елементи підгрупи:", H)
print("Порядок підгрупи:", len(H))
# 5. Самообернені елементи
self_inverse = [x for x in G if x + x == 0]
print("\nСамообернені елементи:")
print(self_inverse)
# 6. Перевірка комутативності
print("\nПеревірка:")
print("a + b =", a + b)
print("b + a =", b + a)
print("Група Z_10 абелева.")
```

```sage
# ГРУПА Z_10^*
R = Integers(10)
G = R.unit_group()
print("Група Z_10^*")
print("Порядок групи:", G.order())
print("Елементи групи:", G.list())
# 1. Вибираємо елементи a і b з абстрактної групи
# G використовує показники твірних, тому беремо саму твірну та її куб.
a = G.gens()[0]
b = a^3
print("\na =", a)
print("b =", b)
# 2. Добуток a * b
print("\na * b =", a * b)
# 3. Обернені елементи
a_inv = a^(-1)
b_inv = b^(-1)
print("\na^(-1) =", a_inv)
print("b^(-1) =", b_inv)
# Перевірка
print("\nПеревірка:")
print("a * a^(-1) =", a * a_inv)
print("b * b^(-1) =", b * b_inv)
# 4. Підгрупа, породжена a і b
H = G.subgroup([a, b])
print("\nПідгрупа <a,b> =", H)
print("Порядок підгрупи:", H.order())
print("Елементи підгрупи:", H.list())
# 5. Самообернені елементи
self_inverse = [x for x in G if x^2 == G.one()]
print("\nСамообернені елементи:")
print(self_inverse)
# 6. Перевірка комутативності
print("\nПеревірка:")
print("a * b =", a * b)
print("b * a =", b * a)
print("a * b == b * a:", a * b == b * a)
print("\nГрупа Z_10^* абелева.")
```

```sage
# ГРУПА A_4
G = AlternatingGroup(4)
print("Група A_4")
print("Порядок групи:", G.order())
print("\nЕлементи групи:")
for x in G:
    print(x)
# 1. Вибираємо елементи a і b
a = G("(1,2,3)")
b = G("(1,2,4)")
print("\na =", a)
print("b =", b)
# 2. Добуток a * b
print("\na * b =", a * b)
# 3. Обернені елементи
a_inv = a^(-1)
b_inv = b^(-1)
print("\na^(-1) =", a_inv)
print("b^(-1) =", b_inv)
# Перевірка
print("\nПеревірка:")
print("a * a^(-1) =", a * a_inv)
print("b * b^(-1) =", b * b_inv)
# 4. Підгрупа, породжена a і b
H = G.subgroup([a, b])
print("\nПідгрупа <a,b>:")
for x in H:
    print(x)
print("Порядок підгрупи:", H.order())
# 5. Самообернені елементи
self_inverse = [x for x in G if x^(-1) == x]
print("\nСамообернені елементи:")
for x in self_inverse:
    print(x)
# 6. Перевірка некомутативності
print("\nПеревірка комутативності:")
print("a * b =", a * b)
print("b * a =", b * a)
print("a * b == b * a:", a * b == b * a)
if a * b != b * a:
    print("A_4 НЕ є абелевою групою.")
```

```sage
# ГРУПА GL(2, Z_3)
F = GF(3)
G = GL(2, F)
print("Група GL(2, Z_3)")
print("Порядок групи:", G.order())
# 1. Вибираємо елементи a і b
a = matrix(F, [
    [1, 1],
    [0, 1]
])
b = matrix(F, [
    [1, 0],
    [1, 1]
])
print("\na =")
print(a)
print("\nb =")
print(b)
# Перевірка, що матриці належать GL(2,Z_3)
print("\ndet(a) =", a.det())
print("det(b) =", b.det())
# 2. Добуток a * b
print("\na * b =")
print(a * b)
# 3. Обернені елементи
a_inv = a^(-1)
b_inv = b^(-1)
print("\na^(-1) =")
print(a_inv)
print("b^(-1) =")
print(b_inv)
# Перевірка
print("\nПеревірка:")
print("a * a^(-1) =")
print(a * a_inv)
print("b * b^(-1) =")
print(b * b_inv)
# 4. Підгрупа, породжена a і b
H = G.subgroup([a, b])
print("\nПорядок підгрупи <a,b>:", H.order())
# 5. Самообернені елементи
self_inverse = [x for x in G if x^2 == G.one()]
print("\nКількість самообернених елементів:", len(self_inverse))
print("Самообернені елементи:")
for x in self_inverse:
    print(x)
# 6. Перевірка некомутативності
print("\nПеревірка комутативності:")
print("a * b =")
print(a * b)
print("b * a =")
print(b * a)
print("a * b == b * a:", a * b == b * a)
if a * b != b * a:
    print("GL(2,Z_3) НЕ є абелевою групою.")
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
# ГРУПА Z_24
G = CyclicPermutationGroup(24)
print("Група Z_24")
print("Порядок групи:", G.order())
# 1. Побудова групи
print("\nЕлементи групи:")
for x in G:
    print(x)
# 2. Усі підгрупи
subgroups = G.subgroups()
print("\nКІЛЬКІСТЬ ПІДГРУП:", len(subgroups))
print("\nУсі підгрупи:")
for i, H in enumerate(subgroups, 1):
    print("H" + str(i) + " =", H)
    print("   порядок =", H.order())
# 3. Підгрупи з точністю до ізоморфізму
classes = []
for H in subgroups:
    found = False
    for C in classes:
        if H.is_isomorphic(C[0]):
            C.append(H)
            found = True
            break
    if not found:
        classes.append([H])
print("\nПІДГРУПИ З ТОЧНІСТЮ ДО ІЗОМОРФІЗМУ:")
for i, C in enumerate(classes, 1):
    H = C[0]
    print("Клас", i)
    print("  Представник:", H)
    print("  Порядок:", H.order())
    print("  Кількість таких підгруп:", len(C))
# 4. Порядки всіх підгруп
print("\nПОРЯДКИ ПІДГРУП:")
for i, H in enumerate(subgroups, 1):
    print("H" + str(i) + ": порядок =", H.order())
# 5. Перевірка теореми Лагранжа
print("\nПЕРЕВІРКА: порядок підгрупи ділить порядок групи")
for H in subgroups:
    print(
        "ord(H) =", H.order(),
        "; 24 mod", H.order(), "=",
        24 % H.order()
    )
print("\nВисновок:")
print("Порядок кожної підгрупи ділить 24.")
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
# ГРАФ КЕЛІ ГРУПИ D_4
G = DihedralGroup(4)
print("Група D_4")
print("Порядок групи:", G.order())
# Система твірних
r = G.gen(0)
s = G.gen(1)
S = [r, s]
print("\nСистема твірних S:")
print("r =", r)
print("s =", s)
# 1. Побудова графа Келі
C = DiGraph()
for g in G:
    C.add_vertex(g)
for g in G:
    for x in S:
        C.add_edge(g, g*x)
print("\nГраф Келі побудовано.")
print("Кількість вершин:", C.num_verts())
print("Кількість орієнтованих ребер:", C.num_edges())
# 2. Степені вершин
print("\nСтепені вершин:")
degrees = []
for g in G:
    out_degree = C.out_degree(g)
    in_degree = C.in_degree(g)
    degrees.append((out_degree, in_degree))
    print(g, ": out-degree =", out_degree,
          ", in-degree =", in_degree)
print("\nУсі вершини мають out-degree =", set(x[0] for x in degrees))
print("Усі вершини мають in-degree =", set(x[1] for x in degrees))
# 3. Вершинно-транзитивність
# Для графа Келі ліві множення
# L_a(g) = a*g є автоморфізмами.
vertex_transitive = True
for a in G:
    for g in G:
        for x in S:
            if a*(g*x) not in G:
                vertex_transitive = False
print("\nЧи є граф вершинно-транзитивним?")
print(vertex_transitive)
# 4. Пошук гамільтонового циклу
vertices = list(G)
start = G.one()
path = [start]
visited = {start}
def find_hamiltonian(path, visited):
    if len(path) == len(vertices):
        if C.has_edge(path[-1], start):
            return path
        return None
    current = path[-1]
    for x in S:
        nxt = current*x
        if nxt not in visited:
            visited.add(nxt)
            path.append(nxt)
            result = find_hamiltonian(path, visited)
            if result is not None:
                return result
            path.pop()
            visited.remove(nxt)
    return None
cycle = find_hamiltonian(path, visited)
print("\nГамільтонів цикл:")
if cycle is not None:
    print(cycle)
    print("Повернення:", cycle[-1], "->", cycle[0])
    print("Гамільтонів цикл існує.")
else:
    print("Гамільтонів цикл не знайдено.")
```
